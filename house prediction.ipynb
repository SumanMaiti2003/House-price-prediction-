```python
x=data.drop(columns=["price"])
y=data['price']
```

```python
x.head()
```

```python
y.head()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression ,Lasso,Ridge
from sklearn.preprocessing import OneHotEncoder,StandardScaler
from sklearn.compose import make_column_transformer
from sklearn.pipeline import make_pipeline
from sklearn.metrics import r2_score
```

```python
import warnings
warnings.filterwarnings('ignore')
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=0)
```

```python
print(x_train.shape)
print(y_train.shape)
```

# Aplyling Linear Regression

```python
column_transform = make_column_transformer(
    (OneHotEncoder(sparse_output=False), ['beds']),
    remainder='passthrough'
)
```

```python
scaler =StandardScaler()
x_scaled=scaler.fit_transform(x)
```

```python
lr=LinearRegression()
lr.fit(x_scaled,y)
```

```python
pipe =make_pipeline(column_transform,scaler,lr)
```

```python
pipe.fit(x_train,y_train)
```

```python
y_pred_lr=pipe.predict(x_test)
```

```python
r2_score(y_test,y_pred_lr)*100
```

## using Lasso

```python
lasso = Lasso() 
```

```python
pipe = make_pipeline(column_transform,scaler,lasso)
```

```python
pipe.fit(x_train,y_train)
```

```python
y_pred_lasso=pipe.predict(x_test)
r2_score(y_test,y_pred_lasso)*100
```

# Using Ridge 

```python
ridge= Ridge()
```

```python
pipe =make_pipeline(column_transform,scaler,ridge)
pipe.fit(x_train,y_train)
```

```python
y_pred_ridge=pipe.predict(x_test)
r2_score(y_test,y_pred_ridge)*100
```

```python
print("No Regularization :",r2_score(y_test,y_pred_lr)*100)
```

```python
print("Lasso :",r2_score(y_test,y_pred_lasso)*100)
```

```python
print("Ridge:",r2_score(y_test,y_pred_ridge)*100)
```
